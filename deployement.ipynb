```python
import pickle
```

```python
# Load both
with open("model/model.pkl", "rb") as f:
    nb_model = pickle.load(f)

with open("model/tfidf_vectorizer.pkl", "rb") as f:
    vectorizer = pickle.load(f)

# Use them for new predictions
text = ['I\'m having trouble getting a job, what should I do to calm myself down?']
X_input = vectorizer.transform(text)
pred = nb_model.predict(X_input)[0]
pred = "is depressed" if pred == 1 else "is not depressed"
print("Predicted class:", pred)
```

```python
import ipywidgets as widgets
from IPython.display import display, clear_output
import numpy as np
```

```python
text_input = widgets.Textarea(value='', placeholder='Type or paste text here...', description='Input:', layout=widgets.Layout(width='100%', height='50px'))
predict_button = widgets.Button(description='🔮 Predict', button_style='success', layout=widgets.Layout(width='150px'))
output = widgets.Output()

def on_predict_clicked(b):
    with output:
        clear_output()
        text = text_input.value.strip()
        if not text:
            print("Please enter some text.")
            return

        # Transform text using your loaded vectorizer
        X_input = vectorizer.transform([text])

        # Predict label
        pred = nb_model.predict(X_input)[0]

        # Convert numeric prediction into readable label
        pred_label = "is depressed" if pred == 1 else "is not depressed"

        # Confidence (probability)
        if hasattr(nb_model, "predict_proba"):
            pred_proba = np.max(nb_model.predict_proba(X_input)) * 100
            print(f"The person {pred_label}")
            print(f"Confidence: {pred_proba:.2f}%")
        else:
            print(f"The person {pred_label}")

predict_button.on_click(on_predict_clicked)

# display UI
display(widgets.VBox([
    widgets.HTML("<h3>🧠 Naive Bayes Text Classifier</h3>"),
    text_input,
    predict_button,
    output
]))
```
